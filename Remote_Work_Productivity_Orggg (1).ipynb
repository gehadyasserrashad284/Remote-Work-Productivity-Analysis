## Phase B — Clean & Transform

This phase prepares the validated dataset for analysis. The required cleaning
and transformation steps are applied to the dataset structure, while keeping
all valid records because no missing values, duplicate rows, duplicate
Employee IDs, or invalid score/percentage values were found.

### B1. Date and Time Transformations

`Survey_Date` is converted to datetime and used to derive survey year, month,
weekday, and quarter.

```python
# Survey_Date is already converted above; derive the required date fields.
df['Survey_Date'] = pd.to_datetime(df['Survey_Date'])

df['survey_year'] = df['Survey_Date'].dt.year
df['survey_month'] = df['Survey_Date'].dt.month
df['survey_weekday'] = df['Survey_Date'].dt.day_name()
df['survey_quarter'] = df['Survey_Date'].dt.quarter

print(df[['Survey_Date', 'survey_year', 'survey_month',
          'survey_weekday', 'survey_quarter']].head())
```

### B2. Meeting-Intensity Bands

Weekly meeting hours are grouped into Low, Medium, and High descriptive
bands for comparison. These are analytical categories, not scientific
thresholds.

```python
df['meeting_intensity'] = pd.cut(
    df['Meeting_Hours_Weekly'],
    bins=[-float('inf'), 5, 10, float('inf')],
    labels=['Low', 'Medium', 'High']
)

print(df['meeting_intensity'].value_counts().sort_index())
```

### B3. Focus-Hour Bands

Daily focus hours are grouped into Low, Medium, and High descriptive bands
to support comparison in the analysis.

```python
df['focus_hour_band'] = pd.cut(
    df['Focus_Hours_Daily'],
    bins=[-float('inf'), 5, 7, float('inf')],
    labels=['Low', 'Medium', 'High']
)

print(df['focus_hour_band'].value_counts().sort_index())
```

### B4. Work-Arrangement Group



The existing work-arrangement categories are standardized and copied into a
dedicated analysis field for consistent grouping.

```python
df['work_arrangement_group'] = (
    df['Work_Arrangement'].astype(str).str.strip().str.title()
)

print(df['work_arrangement_group'].value_counts())
```

###B5. Phase B Validation

Confirm that the required transformed fields were created successfully.

```python
phase_b_columns = [
    'survey_year',
    'survey_month',
    'survey_weekday',
    'survey_quarter',
    'meeting_intensity',
    'focus_hour_band',
    'work_arrangement_group',
]

print("Phase B columns created:")
print(phase_b_columns)

print("\nMissing values in Phase B columns:")
print(df[phase_b_columns].isnull().sum())

print("\nDataset shape after Phase B:", df.shape)
```

Q1 — Work Arrangement Analysis


```python
work_arrangement_analysis = df.groupby('Work_Arrangement').agg(
    Focus=('Focus_Score', 'mean'),
    Collaboration=('Collaboration_Score', 'mean'),
    Work_Life_Balance=('Work_Life_Balance_Score', 'mean'),
    Satisfaction=('Satisfaction_Score', 'mean'),
).round(2)

print(work_arrangement_analysis)
```

Q2 — Department and Job Level Differences

```python
department_analysis = df.groupby('Department').agg(
    Satisfaction=('Satisfaction_Score', 'mean'),
    Burnout=('Burnout_Risk_Percent', 'mean')
).round(2)

print(department_analysis)
```

```python
job_level_analysis = df.groupby('Job_Level').agg(
    Satisfaction=('Satisfaction_Score', 'mean'),
    Burnout=('Burnout_Risk_Percent', 'mean'),
    Tasks_Per_Week=('Tasks_Completed_Weekly', 'mean')
).round(2)

print(job_level_analysis)
```

Q3 — Region Analysis

```python
region_analysis = df.groupby('Region').agg(
    Satisfaction=('Satisfaction_Score', 'mean'),
    Burnout=('Burnout_Risk_Percent', 'mean'),
    Work_Life_Balance=('Work_Life_Balance_Score', 'mean'),
    Tasks_Per_Week=('Tasks_Completed_Weekly', 'mean')
).round(2)

print(region_analysis)
```

```python
company_size_analysis = df.groupby('Company_Size').agg(
    Satisfaction=('Satisfaction_Score', 'mean'),
    Burnout=('Burnout_Risk_Percent', 'mean')
).round(2)

print(company_size_analysis)
```

Q4 — Monthly Outcomes

```python
monthly_analysis = df.groupby(
    ['survey_year', 'survey_month']
).agg(
    Satisfaction=('Satisfaction_Score', 'mean'),
    Burnout=('Burnout_Risk_Percent', 'mean'),
    Focus=('Focus_Score', 'mean'),
    Tasks_Per_Week=('Tasks_Completed_Weekly', 'mean'),
    Responses=('Employee_ID', 'count')
).round(2)

print(monthly_analysis)
```

Q5 Relationships Between Variables


```python
correlations = df[
    [
        'Meeting_Hours_Weekly',
        'Focus_Score',
        'Burnout_Risk_Percent',
        'Commute_Minutes_Daily',
        'Work_Life_Balance_Score',
        'Satisfaction_Score',
        'Focus_Hours_Daily',
        'Tasks_Completed_Weekly'
    ]
].corr()

print("Meeting Hours vs Focus:",
      round(correlations.loc['Meeting_Hours_Weekly', 'Focus_Score'], 2))

print("Meeting Hours vs Burnout:",
      round(correlations.loc['Meeting_Hours_Weekly', 'Burnout_Risk_Percent'], 2))

print("Commute vs Work-Life Balance:",
      round(correlations.loc['Commute_Minutes_Daily', 'Work_Life_Balance_Score'], 2))

print("Commute vs Satisfaction:",
      round(correlations.loc['Commute_Minutes_Daily', 'Satisfaction_Score'], 2))

print("Commute vs Burnout:",
      round(correlations.loc['Commute_Minutes_Daily', 'Burnout_Risk_Percent'], 2))

print("Focus Hours vs Tasks:",
      round(correlations.loc['Focus_Hours_Daily', 'Tasks_Completed_Weekly'], 2))

print("Burnout vs Satisfaction:",
      round(correlations.loc['Burnout_Risk_Percent', 'Satisfaction_Score'], 2))
```

Q6 Overall High-Burnout Share

## Phase E — Data Modeling

### E1. Main Fact Table

The cleaned dataset is represented as the main fact table:

- **Table name:** `cleaned_survey_responses`
- **Grain:** One row per employee survey response
- **Primary Key:** `Employee_ID`
- **Expected rows:** 10,000
- **Final columns:** 26

### E2. Summary Tables

The following summary tables are created from the cleaned fact data:

- `summary_by_work_arrangement`
- `summary_by_department`
- `summary_by_region`
- `summary_by_month`
- `summary_by_company_size_joblevel`

### E3. Relationships

The summary tables are aggregated outputs generated from the main fact table. No artificial parent-child relationships or foreign keys are required between the summary tables.

```python
response_time = (
    df.groupby('Work_Arrangement')['Response_Time_Hours']
    .mean()
    .sort_values()
    .round(2)
)

plt.figure(figsize=(7, 5))
plt.bar(response_time.index, response_time.values)
plt.title('Average Response Time by Work Arrangement')
plt.xlabel('Work Arrangement')
plt.ylabel('Average Response Time (Hours)')
plt.show()
```

```python
fact_table = df.copy()

print("Fact Table:")
print("Table Name: cleaned_survey_responses")
print("Grain: One row per employee survey response")
print("Primary Key: Employee_ID")
print("Rows:", len(fact_table))
print("Columns:", fact_table.shape[1])
```

```python
# Summary table by Work Arrangement
summary_by_work_arrangement = df.groupby('Work_Arrangement').agg(
    Employees=('Employee_ID', 'count'),
    Avg_Satisfaction=('Satisfaction_Score', 'mean'),
    Avg_Burnout=('Burnout_Risk_Percent', 'mean'),
    Avg_Focus=('Focus_Score', 'mean'),
    Avg_Collaboration=('Collaboration_Score', 'mean')
).round(2)

# Summary table by Department
summary_by_department = df.groupby('Department').agg(
    Employees=('Employee_ID', 'count'),
    Avg_Satisfaction=('Satisfaction_Score', 'mean'),
    Avg_Burnout=('Burnout_Risk_Percent', 'mean')
).round(2)

# Summary table by Region
summary_by_region = df.groupby('Region').agg(
    Employees=('Employee_ID', 'count'),
    Avg_Satisfaction=('Satisfaction_Score', 'mean'),
    Avg_Burnout=('Burnout_Risk_Percent', 'mean')
).round(2)

# Summary table by Month
summary_by_month = df.groupby(['survey_year', 'survey_month']).agg(
    Employees=('Employee_ID', 'count'),
    Avg_Satisfaction=('Satisfaction_Score', 'mean'),
    Avg_Burnout=('Burnout_Risk_Percent', 'mean')
).round(2)

# Summary table by Company Size and Job Level
summary_by_company_size_joblevel = df.groupby(
    ['Company_Size', 'Job_Level']
).agg(
    Employees=('Employee_ID', 'count'),
    Avg_Satisfaction=('Satisfaction_Score', 'mean'),
    Avg_Burnout=('Burnout_Risk_Percent', 'mean')
).round(2)

print("Summary tables created successfully.")
```

## Phase I — Data Storytelling

### Key Findings

- Work arrangement is the clearest differentiating factor in the dataset, particularly for burnout risk, focus, and response time.
- Office employees show a substantially higher high-burnout rate than Hybrid and Remote employees.
- Remote employees show stronger average focus, while Office employees show the shortest average response time.
- Department, region, and company-size differences are comparatively small for average satisfaction and burnout.
- Correlation analysis helps identify relationships among meeting time, commute, focus, work-life balance, satisfaction, burnout, and task completion.

### Recommendations

- Consider flexible or hybrid work arrangements where job requirements allow.
- Investigate the factors contributing to higher burnout among Office employees.
- Review meeting practices because meeting time is related to focus and burnout.
- Consider communication expectations separately from well-being outcomes, since response time differs by work arrangement.

### Interpretation Note

These results describe associations in the dataset. They do not prove that work arrangement causes burnout, satisfaction, or productivity changes.

## Phase J — Conclusion and Limitations

### Conclusion

The analysis identifies work arrangement as the strongest visible differentiator in the dataset. Hybrid and Remote groups show more favorable well-being patterns, while Office employees have higher high-burnout rates and faster response times.

### Limitations

- The dataset is cross-sectional and does not track the same employees over time.
